# Deleting redundant columns

```python
data.columns
```

```python
# List of columns that are required for analysis and will be retained in the dataset
columns_to_keep = [
    'name',
    'online_order',
    'book_table',
    'rate',
    'dish_liked',
    'approx_cost(for two people)'
]

# List of columns that are not relevant for the current analysis and will be removed
# Dropping these columns helps reduce noise and optimize the dataset
columns_to_drop = [
    'url',
    'address',
    'votes',
    'phone',
    'location',
    'rest_type',
    'cuisines',
    'reviews_list',
    'menu_item',
    'listed_in(type)',
    'listed_in(city)'
]
```

```python
len(columns_to_keep)
```

```python
len(columns_to_drop)
```

```python
# Check the data type of the 'name' column
# This helps verify whether the column is stored as a Pandas Series
type(data["name"])
```

```python
# Create a new DataFrame by selecting only the required columns
# This helps focus the analysis on relevant data and removes unnecessary fields
modify_data = data[columns_to_keep]
```

```python
modify_data
```

```python
# Drop unnecessary columns from the DataFrame to reduce dataset size
# inplace=True modifies the original DataFrame without creating a copy
data.drop(columns=columns_to_drop, inplace=True)
```

```python
data
```

# Renaming the columns

```python
data.columns
```

```python
# Loop through each column name in the DataFrame
# Capitalize the first letter of each column name and print it
# Useful for quickly reviewing column names in a more readable format
for i in data.columns:
    print(i.capitalize())
```

```python
# Create an empty list to store updated column names
new_column_names = []

# Loop through existing column names in the DataFrame
# Capitalize the first letter of each column name
# Store the transformed column names in a new list
for i in data.columns:
    new_column_names.append(i.capitalize())

```

```python
new_column_names
```

```python
# Assign the new list of column names to the DataFrame
# This updates the DataFrame with standardized, capitalized column names
data.columns = new_column_names
```

```python
data.columns
```

# Dropping duplicates

```python
# Check the total number of duplicate rows in the DataFrame
# Helps identify redundant records that may affect data quality
data.duplicated().sum()
```

```python
# Count how many rows are duplicated and how many are unique
# Useful for understanding the overall duplicate distribution in the dataset
data.duplicated().value_counts()
```

```python
data.duplicated().head(15)
```

```python
data.duplicated().tail(15)
```

```python
# Remove duplicate rows from the DataFrame
# inplace=True ensures duplicates are dropped directly from the original dataset
data.drop_duplicates(inplace=True)
```

```python
# Recheck the number of duplicate rows after removal
# Confirms that duplicate records have been successfully dropped
data.duplicated().sum()
```

# Remove the NaN values from the dataset

```python
# Calculate the total number of missing (NaN) values in each column
# Helps identify columns that require missing value handling
data.isna().sum()
```

```python
# Remove rows containing any missing (NaN) values
# inplace=True updates the DataFrame directly without creating a new copy
data.dropna(inplace=True)
```

```python
data.isna().sum()
```

```python
data
```

```python
# Reset the DataFrame index after row deletions
# This creates a new default integer index
data.reset_index(inplace = True)
```

```python
data
```

```python
# Drop the old index column created during index reset
# Keeps the DataFrame clean without unnecessary columns
data.drop(columns="index", inplace=True)
```

```python
data
```

```python
data["Rate"]
```

```python
# Remove extra spaces from the values in the 'Rate' column
# Updates the column to ensure consistent formatting for analysis
data["Rate"] = data["Rate"].str.replace(" ", "")
```

```python
data
```

```python
# Replace the '/' character with descriptive text in the 'Rate' column
# Improves readability and standardizes the rating format
data["Rate"] = data["Rate"].str.replace("/", " out of ")
```

```python
data
```

# Checking for some more transformations

```python
# Convert the 'Online_order' column from categorical values to binary format
# Assigns 1 if online ordering is available ("Yes"), otherwise assigns 0
# This makes the column suitable for numerical analysis
data["Online_order"] = data["Online_order"].apply(lambda x: 1 if x == "Yes" else 0)
```
